This notebook shows our parameter estimation workflow applied to our kinetic data from all sequences used in this work. the analyzekineticdata() function performs parameter estimation, plots kinetic results, and displays parameter uncertainty intervals.

If you want to test how the initiation-limited model would predict these results, change the fourth argumnent of analyzekineticdata() to 1. 

```julia
#This is just figure one of the main text
flucfilename = "data_processing/output/Fluc/Fluc_PDS_V15_Nathan.csv"
N_all = 2078
N_plasmid = 4352
plt = analyzekineticdata(flucfilename,N_all,N_plasmid,2,showuncertainty = true)
savefig(plt,"./figures/pubfigs/Fluc.png")
savefig(plt,"./figures/pubfigs/Fluc.pdf")
```
