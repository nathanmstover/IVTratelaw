```julia
using Pkg
Pkg.activate("IE")
using DataFrames
using Plots
using XLSX
using ColorSchemes
using Statistics
```

```julia
"""
    importphlabdata()

Import excel file with data from experiments with various conditions

"""

function importphlabdata(filename::String)
    # load excel file into DataFrame
    raw_data = XLSX.readtable(filename, "Sheet1") |> DataFrame
    raw_data_matrix = Matrix(raw_data)

    raw_data_matrix = map(x -> tryparse(Float64, string(x)) !== nothing ? parse(Float64, string(x)) : NaN, raw_data_matrix)

    # create a table with condition number, replicate number, time (hr), mass (kg), NTP conc in Mx4
    processeddata = zeros(size(raw_data_matrix, 1), 8) 

    # fill processed data matrix
    processeddata[:,1] = raw_data_matrix[:,1] # condition number
    processeddata[:,2] = raw_data_matrix[:,2] # replicate number
    processeddata[:,3] = raw_data_matrix[:,5] # Min to hr
    processeddata[:,4] = raw_data_matrix[:,6] .* 0.001 # g to kg
    processeddata[:,5:8] = raw_data_matrix[:,7:10] .* 0.001 # units from mM to M

    processed_data_df = DataFrame(processeddata, :auto)
    processed_data_df= DataFrame(processed_data_df, [:condition, :replicate, :time, :mass, :ATP, :UTP, :CTP, :GTP])
    #remove data point if mass is larger than 7 or less than 5 and remove zero time points measurements
   # filtereddata = filter(row -> 0.005 < row[:mass] < 0.007, processed_data_df)
    
    return filtereddata
end


datafileph = "Data/20240610_FlucExperimentIE_Conditions.xlsx"
filtereddata = importphlabdata(datafileph)
```

```julia
function rate_estimation(NTP::Symbol, filtereddata)
    filtereddata_NTP = select(filteredpHdata, [:condition, :replicate, :time, NTP])

    #find the number of conditions
    filtereddata_NTP_bycond = groupby(filtereddata_NTP, :condition)
    n_cond = length(filtereddata_NTP_bycond)

    #find max number of replicates
    max_repl = 0
    for df_cond in filtereddata_NTP_bycond
        df_cond_byrepl = groupby(df_cond, :replicate)
        num_replicates = length(df_cond_byrepl)
        max_repl = max(max_repl, num_replicates)
    end

    #initialise matrix to store the different y and x values
    y_val_t2 = zeros(n_cond, max_repl)
    y_val_t5 = zeros(n_cond, max_repl) 
    x_val_t2 = zeros(n_cond, max_repl)
    x_val_t5 = zeros(n_cond, max_repl)

    #initialise matrix to delta x and delta y
    dy_val =  zeros(n_cond, max_repl)
    dx_val =  zeros( n_cond, max_repl) 
    
    #initialise matrix to store average delta x and delta y across replicates
    dy_val_avg =  zeros(n_cond,1)
    dx_val_avg =  zeros(n_cond,1) 

    #initialise matrix to store rates for each condition
    rate =  zeros(n_cond, 1) #[M/hr]

    for (i, df_cond) in enumerate(filtereddata_NTP_bycond)
        df_cond_byrepl = groupby(df_cond, :replicate)

        for (j, df_repl) in enumerate(df_cond_byrepl)
            if nrow(df_repl) < 4
                continue
            end       
            #store the concentration at time 2 and 5
            df_repl_time_M = select(df_repl, :time, NTP)
            y_val_t2[i,j] = df_repl_time_M[1,2] #row 1 because already deleted time 1
            y_val_t5[i,j] = df_repl_time_M[4,2] 
            dy_val[i,j] = y_val_t5[i,j] - y_val_t2[i,j] 
            #store the time value at time 2 (x)
            x_val_t2[i,j] = df_repl_time_M[1,1]
            x_val_t5[i,j] = df_repl_time_M[4,1]
            dx_val[i,j] = x_val_t5[i,j] - x_val_t2[i,j] 
        end
    dy_nonzero =  dy_val[i, dy_val[i, :] .!= 0]
    dx_nonzero = dx_val[i, dx_val[i, :] .!= 0]
        
    #average y and x values
    if length(dy_nonzero) > 0
            dy_val_avg[i, 1] = mean(dy_nonzero) # average the delta values across replicates ##might have to update later with weighting more/less some of the replicates
        else
            dy_val_avg[i, 1] = 0
        end

        if length(dx_nonzero) > 0
            dx_val_avg[i, 1] = mean(dx_nonzero) # idem above
        else
            dx_val_avg[i, 1] = 0
        end

        # calculate rate
        if dx_val_avg[i, 1] != 0
            rate[i, 1] = dy_val_avg[i, 1] / dx_val_avg[i, 1] * 4 * 60  # 4 is for the 4 NTPs, 60 to convert min to hour
        else
            rate[i, 1] = 0
        end
    end
    
    return rate
end      
    
```
