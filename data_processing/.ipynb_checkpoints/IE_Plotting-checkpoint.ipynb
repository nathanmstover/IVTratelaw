```julia
"""
    plotrawdata(NTP::Symbol, datafile)

Generate plot of the experimental data of multiple replicates for each condition
"""
function plotrawdata(NTP::Symbol, datafile::String)
    df = importlabdata(datafile) 

    # select the specified NTP data
    df_NTP = select(df, [:condition, :replicate, :time, NTP])

    df_NTP_bycond = groupby(df_NTP, :condition)
    n_conditions = length(df_NTP_bycond)

    # calculate index of subplots
    nrows = ceil(Int, n_conditions / 2)
    ncols = 2

    # create plotting object
    p = plot(layout=(nrows, ncols), size=(1000, 1350))

    subplot_index = 1

    # iterate over conditions
    for (i, df_i) in enumerate(df_NTP_bycond)
        df_NTP_bycond_byrepl = groupby(df_i, :replicate)

        # iterate over replicates within each condition
        for (j, df_j) in enumerate(df_NTP_bycond_byrepl)
            plot!(p[subplot_index], df_j.time, df_j[!, NTP], label="Replicate $j")
        end

        xlabel!("Time (hr)", subplot=subplot_index)
        ylabel!("NTP (M)", subplot=subplot_index)
        title!("Condition: $(unique(df_i.condition)[1])", subplot=subplot_index)

        subplot_index += 1
    end

    display(p)
end
```
