```julia
"""
    importlabdata()

Import excel file with data from experiments with various conditions

"""
function importlabdata(filename::String)
    # load excel file into DataFrame
    raw_data = XLSX.readtable(filename, "Sheet1") |> DataFrame
    raw_data_matrix = Matrix(raw_data)

    raw_data_matrix = map(x -> tryparse(Float64, string(x)) !== nothing ? parse(Float64, string(x)) : NaN, raw_data_matrix)

    # create a table with condition number, replicate number, time (hr), mass (kg), NTP conc in Mx4
    processeddata = zeros(size(raw_data_matrix, 1), 8) 

    # fill processed data matrix
    processeddata[:,1] = raw_data_matrix[:,1] # condition number
    processeddata[:,2] = raw_data_matrix[:,2] # replicate number
    processeddata[:,3] = raw_data_matrix[:,5] ./60 # Min to hr
    processeddata[:,4] = raw_data_matrix[:,6] .* 0.001 # g to kg
    processeddata[:,5:8] = raw_data_matrix[:,7:10] #.* 0.001 # units from mM to M

    processed_data_df = DataFrame(processeddata, :auto)
    processed_data_df= DataFrame(processed_data_df, [:condition, :replicate, :time, :mass, :ATP, :UTP, :CTP, :GTP])
    
    return processed_data_df
end
```

```julia
#import lab data
datafileivt = "Data/FlucExperimentIE_Conditions.xlsx"
filtereddata = importlabdata(datafileivt)
```

```julia
"""
    rate_estimation(data::DataFrame, sigma::Float64)

Calculates the reaction rate for the provided experimental data at a certain condition (which could include replicates) and the measurement error

"""

function rate_estimation(data::DataFrame, sigma::Float64)
    #data should include column with replicate number, time [hr], and preferred (ATP) concentration [M], including headers
    #sigma is the standard devition for all observations in the given data set

    #separate data into df per replicate
    df_byrepl = groupby(data, :replicate)
 
    #find number of replicates
    n_repl = length(df_byrepl)

    #initialise storage df with rate and uncertainty
    rate_error_all_repl = zeros(n_repl, 2) #column 1 = rate, column 2 = uncertainty
    
    #iterate over different replicates
    for (j, df_repl) in enumerate(df_byrepl)
        n_times = length(df_repl.time)                              # number of timepoints in that replicate
        X = ones(n_times, 2)                                        # construct matrix X = [1 timepoints]
        X[:, 2] = df_repl.time                                      # replace the 2nd column with time points
        phi = sigma.^2*I(n_times)           
        Y = df_repl.ATP                                             # actual results are the NTP concentrations   
        
        new_matrix = transpose(X)*inv(phi)*X
        if det(new_matrix) < 10^(-3)
                print("not invertable")
        else
            B = inv(transpose(X)*inv(phi)*X)*transpose(X)*inv(phi)*Y    # find B0 and B1 -> Y = B0 + B1*X
            
            rate_error_all_repl[j,1] = B[2]                             # store B1 (eg slope)
        
            cov = inv(transpose(X)*inv(phi)*X)                          # calculate covariance of b
            times = df_repl.time
            variance = [[x,1]'*cov*[x,1] for x in times]                # calculate the variance 
            variance_average = mean(variance)
            rate_error_all_repl[j,2] = sqrt(variance_average)           # store the variance of the replicate
    
        end
    end

    #combine the data for all replicates for that specific condition
    numerator = 0
    denominator = 0
    for j in 1:n_repl
        numerator += rate_error_all_repl[j, 1] / (rate_error_all_repl[j, 2]^2)
        denominator += 1 / (rate_error_all_repl[j, 2]^2)
    end
    combined_rate = numerator / denominator * 4
    combined_variances = 1/denominator 

    result = [combined_rate, combined_variances]
    
    return result
end      
```

```julia
function filter_last_zero(filtereddata)
    filtered_data = filter(col -> col[end] != 0, filtereddata)
    return filtered_data
end

# Call the function to get filtered data
filtered_data = filter_last_zero(filtereddata)
```

```julia
#make Df of ATP data
filtereddata_ATP = select(filtered_data, [:condition, :replicate, :time, :ATP])

#split up by different conditions
filtereddata_ATP_bycond = groupby(filtereddata_ATP, :condition)

# test data
for df_cond in filtereddata_ATP_bycond
    replicate1 = df_cond[:,2]
    time1 = df_cond[:,3]
    ATP1 = df_cond[:,4]
        
# create dataframe w heading
    data = DataFrame(replicate = replicate1, time = time1, ATP = ATP1)
    
# Display the DataFrame
    rate_estimation(data, 0.5)
    copypaste = -rate_estimation(data, 0.5)[1], rate_estimation(data, 0.5)[2]
    println(copypaste)
end 

```
