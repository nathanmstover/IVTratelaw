```julia
"""
    importlabdata()

Import excel file with data from experiments with various conditions

"""
function importlabdata(filename::String)
    # load excel file into DataFrame
    raw_data = XLSX.readtable(filename, "Sheet1") |> DataFrame
    raw_data_matrix = Matrix(raw_data)
    
    raw_data_matrix = map(x -> tryparse(Float64, string(x)) !== nothing ? parse(Float64, string(x)) : NaN, raw_data_matrix)

    # create a table with condition number, replicate number, time (hr), mass (kg), NTP conc in Mx4
    processeddata = zeros(size(raw_data_matrix, 1), 8) 

    # fill processed data matrix
    processeddata[:,1] = raw_data_matrix[:,1] # condition number
    processeddata[:,2] = raw_data_matrix[:,2] # replicate number
    processeddata[:,3] = raw_data_matrix[:,5] ./60 # Min to hr
    processeddata[:,4] = raw_data_matrix[:,6] .* 0.001 # g to kg
    processeddata[:,5:8] = raw_data_matrix[:,7:10] #.* 0.001 # units from mM to M

    processed_data_df = DataFrame(processeddata, :auto)
    processed_data_df= DataFrame(processed_data_df, [:condition, :replicate, :time, :mass, :ATP, :UTP, :CTP, :GTP])
    
    return processed_data_df
end
```

```julia
function copydata(filename::String)
    # load excel file into DataFrame
    excel_data = XLSX.readtable(filename, "Sheet1") |> DataFrame
    excel_data_matrix = Matrix(excel_data)
    
    excel_data_matrix = map(x -> tryparse(Float64, string(x)) !== nothing ? parse(Float64, string(x)) : NaN, excel_data_matrix)

    # create a table with condition number, replicate number, time (hr), mass (kg), NTP conc in Mx4
    condata = zeros(size(excel_data_matrix, 1), 3) 

    # fill processed data matrix
    condata[:,1] = excel_data_matrix[:,1] # condition
    condata[:,2] = excel_data_matrix[:,3] # DNA inputs
    condata[:,3] = excel_data_matrix[:,4] # T7 inputs
    
    con_data_df = DataFrame(condata, :auto)
    con_data_df= DataFrame(con_data_df, [:condition, :DNAinputs, :T7inputs])
    
    return con_data_df
end
```

```julia
"""
    rate_estimation(data::DataFrame, sigma::Float64)

Calculates the reaction rate for the provided experimental data at a certain condition (which could include replicates) and the measurement error

"""

function rate_estimation(data::DataFrame, sigma::Float64)
    #data should include column with replicate number, time [hr], and preferred (ATP) concentration [M], including headers
    #sigma is the standard deviation for all observations in the given data set

    #separate data into df per replicate
    df_byrepl = groupby(data, :replicate)
 
    #find number of replicates
    n_repl = length(df_byrepl)

    #initialise storage df with rate and uncertainty
    rate_error_all_repl = zeros(n_repl, 2) #column 1 = rate, column 2 = uncertainty
    
    #iterate over different replicates
    for (j, df_repl) in enumerate(df_byrepl)
        n_times = length(df_repl.time)                              # number of timepoints in that replicate
        X = ones(n_times, 2)                                        # construct matrix X = [1 timepoints]
        X[:, 2] = df_repl.time                                      # replace the 2nd column with time points
        phi = sigma.^2*I(n_times)                                   #[sigma] = mM [phi] = mM^2
        Y = df_repl.ATP    #[Y] =  mM                               # actual results are the NTP concentrations   
        
        new_matrix = transpose(X)*inv(phi)*X
        if det(new_matrix) < 10^(-3)
                print("not invertable")
        else
            B = inv(transpose(X)*inv(phi)*X)*transpose(X)*inv(phi)*Y # find B0 and B1 -> Y = B0 + B1*X
                                                                      #[B] = mM/hr 
            rate_error_all_repl[j,1] = B[2]                          # store B1 (eg slope)
        
            cov = inv(transpose(X)*inv(phi)*X)                       # calculate covariance of b #[cov]=mM^2/hr^2
            times = df_repl.time
            variance = [[x,1]'* cov*[x,1] for x in times]            # calculate the variance #[var]=mM^2/hr^2
            variance_average = sum(variance) #mean(variance)   #average across time points
            rate_error_all_repl[j,2] = sqrt(variance_average)        # store the st dev of the replicate
    
        end
    end

    #combine the data for all replicates for that specific condition
    numerator = 0
    denominator = 0
    for j in 1:n_repl
        numerator += rate_error_all_repl[j, 1] / (rate_error_all_repl[j, 2].^2) #1/mM/hr
        denominator += 1 / (rate_error_all_repl[j, 2].^2) #1/mM^2/hr^2
    end
    combined_rate = numerator / denominator * 4   
    combined_variances = 1 / denominator 

    result = [combined_rate, combined_variances]
    
    return result
end      
```

```julia
"""
    rate_estimation(data::DataFrame, sigma::Float64)

Calculates the reaction rate for the provided experimental data at a certain condition (which could include replicates) and the measurement error

"""

function rate_estimation_diff_vari_time(data::DataFrame, sigma::Float64)
    #data should include column with replicate number, time [hr], and preferred (ATP) concentration [M], including headers
    #sigma is the standard deviation for all observations in the given data set

    #separate data into df per replicate
    df_byrepl = groupby(data, :replicate)
 
    #find number of replicates
    n_repl = length(df_byrepl)

    #initialise storage df with rate and uncertainty
    rate_error_all_repl = zeros(n_repl, 2) #column 1 = rate, column 2 = uncertainty

    variance_stored = []
    
    #iterate over different replicates
    for (j, df_repl) in enumerate(df_byrepl)
        n_times = length(df_repl.time)                              # number of timepoints in that replicate
        X = ones(n_times, 2)                                        # construct matrix X = [1 timepoints]
        X[:, 2] = df_repl.time                                      # replace the 2nd column with time points
        phi = sigma.^2*I(n_times)                                   #[sigma] = mM [phi] = mM^2
        Y = df_repl.ATP    #[Y] =  mM                               # actual results are the NTP concentrations   
        
        new_matrix = transpose(X)*inv(phi)*X
        if det(new_matrix) < 10^(-3)
                print("not invertable")
        else
            B = inv(transpose(X)*inv(phi)*X)*transpose(X)*inv(phi)*Y # find B0 and B1 -> Y = B0 + B1*X
                                                                      #[B] = mM/hr 
            rate_error_all_repl[j,1] = B[2]                          # store B1 (eg slope)
        
            cov = inv(transpose(X)*inv(phi)*X)                       # calculate covariance of b #[cov]=mM^2/hr^2
            times = df_repl.time
            variance = [[x,1]'* cov*[x,1] for x in times]            # calculate the variance #[var]=mM^2/hr^2

            #store the variances to see whats happening
            push!(variance_stored, variance)

            weights = 1 ./ variance
            weighted_mean = sum(weights.*times)/sum(weights)
            SS_xw = sum(weights.*(times.-weighted_mean).^2) #weighted sum of squares of deviations of time points from weighted mean
            weighted_residuals = Y .- (B[1] .+ B[2].*times)
            SS_rw = sum(weights.*weighted_residuals.^2)
            variances_w = SS_rw / (sum(weights)-2)
            variance_average = variances_w/SS_xw
            
            
            #variance_average = sum(variance) #mean(variance)   #average across time points
            rate_error_all_repl[j,2] = sqrt(variance_average)        # store the st dev of the replicate
    
        end
    end

    #combine the data for all replicates for that specific condition
    numerator = 0
    denominator = 0
    for j in 1:n_repl
        numerator += rate_error_all_repl[j, 1] / (rate_error_all_repl[j, 2].^2) #1/mM/hr
        denominator += 1 / (rate_error_all_repl[j, 2].^2) #1/mM^2/hr^2
    end
    combined_rate = numerator / denominator * 4   
    combined_variances = 1 / denominator 

    result = [combined_rate, combined_variances]
    
    return result, variance_stored
end      
```

```julia
function filter_last_zero(filtereddata)
    filtered_data = filter(col -> col[end] != 0, filtereddata)
    return filtered_data
end

# Call the function to get filtered data
filtered_data = filter_last_zero(filtereddata)
```

```julia
using XLSX
using Dates  # For date formatting

function adjustExcel(filename::AbstractString, data::Matrix, start_row::Int, start_col::Int, new_filename::AbstractString)
    if isfile(filename)
        # Rename existing file with timestamp
        timestamp = Dates.format(now(), "yy-mm-dd")
        renamed_filename = replace(filename, ".xlsx" => "_$timestamp.xlsx")
        mv(filename, renamed_filename, force=true)
        println("Existing file renamed to: $renamed_filename")
    end
    
    # Open an existing workbook or create a new one
    wb = XLSX.openxlsx(filename, mode="w")
    
    # Select the first sheet in the workbook
    sheet = wb[1]
    
    # Write data to specific cells starting from start_row, start_col
    rows, cols = size(data)
    for r in 1:rows
        for c in 1:cols
            XLSX.setdata!(sheet, start_row + r - 1, start_col + c - 1, data[r, c])
        end
    end
    
    # Save and close the workbook
    XLSX.writexlsx(filename, wb)
end
```

```julia
data_to_write = sorted_matrix # Example data matrix
excel_filename = "FlucIE_ProcessedData_20240625.xlsx"  # Name of the Excel file to write to
start_row = 1 # Starting row in Excel
start_col = 1  # Starting column in Excel
new_filename = "FlucIE_ProcessedData.xlsx" # New sheet name

adjustExcel(excel_filename, data_to_write, start_row, start_col, new_filename)
```
